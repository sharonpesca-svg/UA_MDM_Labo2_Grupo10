```python
# Inicializar BorutaShap
# 'classification' para clasificación, 'regression' para regresión
feature_selector = BorutaShap(model=modelo1, 
                              importance_measure='shap', 
                              classification=True)
```

```python
# Ajustar el selector
# n_trials: Número de iteraciones (min. 50, 100 mejor)
feature_selector.fit(X=X1_train, y=y_serie_train, n_trials=1, sample=11994 , verbose=True)
#, sample=False
```
